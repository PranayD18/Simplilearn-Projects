```python
def find_customers(visits: pd.DataFrame, transactions: pd.DataFrame) -> pd.DataFrame:
    df1=pd.merge(left=visits,right=transactions,how='left',on='visit_id')
    fc=df1['transaction_id'].isnull()
    df1=df1[fc][['visit_id','customer_id']]
    df2=df1.groupby('customer_id').agg({'visit_id':'count'}).reset_index()
    df2=df2.rename(columns = {"visit_id" : "count_no_trans"}).sort_values(by='count_no_trans')
    return df2

find_customers(visits,transactions)
```

```python
def students_and_examinations(students: pd.DataFrame, subjects: pd.DataFrame, examinations: pd.DataFrame) -> pd.DataFrame:
    df1=students
    df2=subjects
    df3=examinations
    df3['dummy']=1
    df4=pd.merge(left=df1, right=df2,how='cross')
    df5=pd.merge(left=df4,right=df3,how='left',on=['student_id','subject_name'])
    df5=df5.groupby(['student_id','student_name','subject_name'],dropna=False).agg({'dummy':'count'}).reset_index()
    df5=df5.rename(columns={'dummy':'attended_exams'})
    return df5
```

```python
import pandas as pd
from numpy import where

def average_selling_price(prices: pd.DataFrame, 
                      units_sold: pd.DataFrame) -> pd.DataFrame:

    df = prices.merge(units_sold, how = 'left')                 # <-- 1)
    df['rev'] = df.price * df.units

    df = (df[(df.purchase_date.isna())|                         # <-- 2)
         (df.purchase_date.between(df.start_date, df.end_date))])

    grp = df.groupby('product_id').agg(                         # <-- 3)
                    total_units = ('units', 'sum'),
                      total_rev = ('rev', 'sum')
                    ).reset_index()

    grp['average_price'] = (where(grp.total_units == 0, 0,      # <-- 4)
                     (grp.total_rev/grp.total_units).round(2)))

    return grp.iloc[:,[0,3]]
```

```python
import pandas as pd

def average_selling_price(prices: pd.DataFrame, units_sold: pd.DataFrame) -> pd.DataFrame:
    units_sold = units_sold.merge(prices, on='product_id', how='left')
    units_sold['matching'] = (units_sold['purchase_date'] >= units_sold['start_date']) & (units_sold['purchase_date'] <= units_sold['end_date'])
    units_sold = units_sold[units_sold['matching']==True]
    units_sold['total_spent'] = units_sold['units']*units_sold['price']
    units_sold = units_sold.groupby('product_id').agg(total_spent=('total_spent', lambda x: sum(x)), total_units=('units', lambda x: sum(x))).reset_index()
    units_sold['average_price'] = round(units_sold['total_spent']/units_sold['total_units'],2)
    prices = pd.DataFrame(columns=['product_id'], data=prices['product_id'].unique())
    return prices.merge(units_sold[['product_id', 'average_price']], on='product_id', how='left').fillna(0)

average_selling_price(prices,units_sold)
```

```python
def monthly_transactions(transactions: pd.DataFrame) -> pd.DataFrame:
    df1=transactions
    df1['year_month'] = df1['trans_date'].dt.to_period('M').astype(str)
    df1['stateCnt']=1
    df1['approved_trans']=np.where(df1['state']=='approved',1,0)
    df1['approved_amount']=np.where(df1['state']=='approved',df1['amount'],0)
    df2=df1.groupby(['year_month','country'],dropna=False).agg({'id':'count','approved_trans':'sum','amount':'sum','approved_amount':'sum'})
    df2=df2.reset_index().rename(columns={'year_month':'month','id':'trans_count','approved_trans':'approved_count','amount':'trans_total_amount','approved_amount':'approved_total_amount'})
    return df2

monthly_transactions(transactions)
```

```python
def user_activity(activity: pd.DataFrame) -> pd.DataFrame:
    end_date=pd.to_datetime('2019-07-27')
    start_date=end_date - pd.DateOffset(30)
    activity['action']=activity['activity_date'].between(start_date,end_date)
    activity=activity[activity['action']==True]
    activity=activity.groupby('activity_date').agg({'user_id':'nunique'})
    activity=activity.rename(columns={'activity_date':'day', 'user_id':'active_users'}).reset_index()
    return activity

user_activity(activity)
```

```python
## Write a solution to report the customer ids from the Customer table that bought all the products in the Product table.

data = [[1, 5], [2, 6], [3, 5], [3, 6], [1, 6]]
customer = pd.DataFrame(data, columns=['customer_id', 'product_key']).astype({'customer_id':'Int64', 'product_key':'Int64'})
data = [[5], [6]]
product = pd.DataFrame(data, columns=['product_key']).astype({'product_key':'Int64'})
```

```python
product['cnt']=len(product['product_key'].drop_duplicates())
product

customer_fc=customer.groupby('customer_id').agg({'product_key':'nunique'}).reset_index().rename(columns={'product_key':'prod_cnt'})
customer_fc
```

```python
df=pd.merge(customer_fc,product,how='inner',left_on='prod_cnt',right_on='cnt')
df=df['customer_id'].drop_duplicates().reset_index()
df
```

```python
def find_customers(customer: pd.DataFrame, product: pd.DataFrame) -> pd.DataFrame:
    product['cnt']=len(product['product_key'].drop_duplicates())
    customer_fc=customer.groupby('customer_id').agg({'product_key':'nunique'}).reset_index().rename(columns={'product_key':'prod_cnt'})
    df=pd.merge(customer_fc,product,how='inner',left_on='prod_cnt',right_on='cnt')
    df=df[['customer_id']].drop_duplicates()
    return df

find_customers(customer, product)
```
